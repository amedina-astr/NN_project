```python
##### VIEW RECOMENDED CLOUDS USING VIRGA

def check_recommended_clouds(bd, mh, mmw, plot=True):
    """
    Call virga.recommend_gas on the current TP profile inside bd.
    """
    prof = bd.inputs['atmosphere']['profile']
    P = np.asarray(prof['pressure'], float)
    T = np.asarray(prof['temperature'], float)

    # This will show you what virga thinks is relevant
    rec = vj.recommend_gas(P, T, mh, mmw, plot=plot, legend='inside')
    return rec
```
