```python
##### IMPORTS
import numpy as np
import matplotlib.pyplot as plt
```

```python
##### CONFIGURATION and BASICS

file_path = r"C:\Users\Alex\Desktop\Picaso\NN_project\sample_out\t1253.0270586044937g1082.0568603568015.npy"

load_file = np.load(file_path, allow_pickle=True)

shape     = load_file.shape
print(shape)

# The shape is (20472,) which means it's a 1D array with 20472 elements
# Makes sense because Peter's thesis, a resolving power of R = 5000
# With a wavelength range from 0.5 to 30 micron corresponds to 20472 elements

# Assuming this is the way
# N = R * ln(lamba_max / lambda_min) + 1

lambda_min, lambda_max = 0.5, 30.0

# Total number of points
N = load_file.size
```

```python
##### BUILD

# Since there is no wavelength, MUST build a wavelength space
# But we know min, max, and the number of points

flux = load_file
wavpoints = np.linspace(lambda_min, lambda_max, N)
print(len(wavpoints))
```

```python
##### PLOT

plt.figure(figsize=(8,5))
plt.plot(wavpoints, flux)
plt.xlabel("Wavelength (micron)")
plt.ylabel("Flux (erg cm^-2 s^-1 cm^-1)")
plt.title("Test Plot (Peter Data)")
plt.show()
```
